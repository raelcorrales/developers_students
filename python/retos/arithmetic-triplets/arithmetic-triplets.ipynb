```python
import timeit
```

```python
def solution(nums: list, diff: int) -> int:
    num_set = set(nums)
    return sum(1 for a in nums if (a + diff in num_set and a + 2 * diff in num_set))
```

```python
execution_time = timeit.timeit(lambda: solution([1, 2, 4, 6, 7, 11, 12], 5), number=10000)
print("Resultado:", solution([1, 2, 4, 6, 7, 11, 12], 5))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

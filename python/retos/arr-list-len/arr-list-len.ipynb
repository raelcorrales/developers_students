```python
import timeit
```

```python
def solution(A: list) -> int:
    length = 0
    curr = 0
    while curr != -1:
        length += 1
        curr = A[curr]
    return length
```

```python
execution_time = timeit.timeit(lambda: solution([1, 4, -1, 3, 2]), number=10000)
print("Resultado:", solution([1, 4, -1, 3, 2]))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
import timeit
```

```python
def solution(n: int) -> int:
    steps = 0
    while n > 1:
        n = n // 2 if n % 2 == 0 else 3 * n + 1
        steps += 1
    return steps
```

```python
execution_time = timeit.timeit(lambda: solution(3), number=10000)
print("Resultado:", solution(3))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

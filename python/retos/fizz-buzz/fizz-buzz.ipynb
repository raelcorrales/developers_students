```python
import timeit # Agregamos timeit para mostrar como es que cada funcion tarda al procesar.
```

```python
def solution(n: int) -> list:
    """
    Genera la secuencia FizzBuzz de 1 a n usando condicionales if/elif/else.
    
    :param n: Límite superior entero positivo.
    :return: Lista de strings con la secuencia.
    """
    result = []
    for num in range(1, n + 1):
        # Paso 1: Evaluamos primero la condición combinada (múltiplo de 3 y 5 = 15).
        if num % 15 == 0:
            result.append("FizzBuzz")
        elif num % 3 == 0:
            result.append("Fizz")
        elif num % 5 == 0:
            result.append("Buzz")
        else:
            result.append(str(num))
    return result
```

```python
def solution_pythonic(n: int) -> list:
    """
    Genera la secuencia FizzBuzz utilizando un mapeo de reglas extensible.
    
    :param n: Límite superior entero positivo.
    :return: Lista de strings con la secuencia.
    """
    # Paso 1: Reglas definidas como pares (divisor, palabra).
    rules = ((3, "Fizz"), (5, "Buzz"))
    result = []
    
    # Paso 2: Iteramos sobre el rango concatenando palabras si el residuo es cero.
    for num in range(1, n + 1):
        output = "".join(word for divisor, word in rules if num % divisor == 0)
        result.append(output or str(num))
        
    return result
```

```python
execution_time = timeit.timeit(lambda: solution(15), number=10000)
print("Resultado:", solution(15))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
execution_time = timeit.timeit(lambda: solution_pythonic(15), number=10000)
print("Resultado:", solution_pythonic(15))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
import timeit
```

```python
def solution(E: str, L: str) -> int:
    t_in = int(E[:2]) * 60 + int(E[3:])
    t_out = int(L[:2]) * 60 + int(L[3:])
    hours = (t_out - t_in + 59) // 60
    return 2 + 3 + max(0, hours - 1) * 4
```

```python
execution_time = timeit.timeit(lambda: solution("10:00", "13:21"), number=10000)
print("Resultado:", solution("10:00", "13:21"))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
import timeit
```

```python
def solution(a: list, b: list) -> list:
    set_b = set(b)
    result = []
    seen = set()
    for item in a:
        if item in set_b and item not in seen:
            result.append(item)
            seen.add(item)
    return result
```

```python
def solution_pythonic(a: list, b: list) -> list:
    return list(set(a) & set(b))
```

```python
execution_time = timeit.timeit(lambda: solution([1, 2, 3, 4, 5], [0, 1, 3, 7]), number=10000)
print("Resultado:", solution([1, 2, 3, 4, 5], [0, 1, 3, 7]))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
execution_time = timeit.timeit(lambda: solution_pythonic([1, 2, 3, 4, 5], [0, 1, 3, 7]), number=10000)
print("Resultado:", solution_pythonic([1, 2, 3, 4, 5], [0, 1, 3, 7]))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

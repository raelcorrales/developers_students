```python
import timeit
import re
```

```python
def solution(S: str) -> int:
    valid_lengths = [
        len(w) for w in S.split()
        if w.isalnum()
        and sum(1 for c in w if c.isalpha()) % 2 == 0
        and sum(1 for c in w if c.isdigit()) % 2 == 1
    ]
    return max(valid_lengths, default=-1)
```

```python
execution_time = timeit.timeit(lambda: solution("test 5 a0A pass007 ?xy1"), number=10000)
print("Resultado:", solution("test 5 a0A pass007 ?xy1"))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

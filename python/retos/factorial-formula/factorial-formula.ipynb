```python
import timeit
import math
```

```python
def solution(n: int) -> int:
    ans = 1
    for i in range(2, n + 1):
        ans *= i
    return ans
```

```python
def solution_pythonic(n: int) -> int:
    return math.factorial(n)
```

```python
execution_time = timeit.timeit(lambda: solution(10), number=10000)
print("Resultado:", solution(10))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
execution_time = timeit.timeit(lambda: solution_pythonic(10), number=10000)
print("Resultado:", solution_pythonic(10))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
import timeit
```

```python
def solution(s: str) -> int:
    rom = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
    total, i, n = 0, 0, len(s)
    while i < n:
        if i + 1 < n and rom[s[i + 1]] > rom[s[i]]:
            total += rom[s[i + 1]] - rom[s[i]]
            i += 2
        else:
            total += rom[s[i]]
            i += 1
    return total
```

```python
def solution_pythonic(s: str) -> int:
    rom = {'I': 1, 'V': 5, 'X': 10, 'L': 50, 'C': 100, 'D': 500, 'M': 1000}
    total, prev = 0, 0
    for char in reversed(s):
        curr = rom[char]
        total = total - curr if curr < prev else total + curr
        prev = curr
    return total
```

```python
execution_time = timeit.timeit(lambda: solution("MCMXCIV"), number=10000)
print("Resultado:", solution("MCMXCIV"))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
execution_time = timeit.timeit(lambda: solution_pythonic("MCMXCIV"), number=10000)
print("Resultado:", solution_pythonic("MCMXCIV"))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
import timeit # Medición de tiempos de ejecución
```

```python
def solution(nums: list) -> bool:
    seen = set()
    for num in nums:
        if num in seen:
            return True
        seen.add(num)
    return False
```

```python
def solution_pythonic(nums: list) -> bool:
    return len(set(nums)) < len(nums)
```

```python
execution_time = timeit.timeit(lambda: solution([1, 2, 3, 4, 5, 2]), number=10000)
print("Resultado:", solution([1, 2, 3, 4, 5, 2]))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
execution_time = timeit.timeit(lambda: solution_pythonic([1, 2, 3, 4, 5, 2]), number=10000)
print("Resultado:", solution_pythonic([1, 2, 3, 4, 5, 2]))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
import timeit
```

```python
def solution(target: int) -> int:
    total = 0
    for i in range(1, target):
        if i % 3 == 0 or i % 5 == 0:
            total += i
    return total
```

```python
def solution_pythonic(target: int) -> int:
    return sum(i for i in range(1, target) if i % 3 == 0 or i % 5 == 0)
```

```python
execution_time = timeit.timeit(lambda: solution(100), number=10000)
print("Resultado:", solution(100))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
execution_time = timeit.timeit(lambda: solution_pythonic(100), number=10000)
print("Resultado:", solution_pythonic(100))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

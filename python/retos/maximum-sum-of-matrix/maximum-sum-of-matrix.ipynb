```python
import timeit
from itertools import permutations
```

```python
def solution(matrix: list) -> int:
    n = len(matrix)
    if not matrix:
        return 0
    return max(sum(matrix[i][p[i]] for i in range(n)) for p in permutations(range(n)))
```

```python
matrix = [[3, 1, 2], [4, 5, 6], [-7, 8, 9]]
execution_time = timeit.timeit(lambda: solution(matrix), number=10000)
print("Resultado:", solution(matrix))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
import timeit
import calendar
```

```python
def solution(year: int) -> bool:
    if year % 400 == 0:
        return True
    elif year % 100 == 0:
        return False
    elif year % 4 == 0:
        return True
    return False
```

```python
def solution_pythonic(year: int) -> bool:
    return calendar.isleap(year)
```

```python
execution_time = timeit.timeit(lambda: solution(2024), number=10000)
print("Resultado:", solution(2024))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

```python
execution_time = timeit.timeit(lambda: solution_pythonic(2024), number=10000)
print("Resultado:", solution_pythonic(2024))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```

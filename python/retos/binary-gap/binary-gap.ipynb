```python
import timeit
```

```python
def solution(N: int) -> int:
    gaps = bin(N)[2:].strip('0').split('1')
    return max((len(gap) for gap in gaps), default=0)
```

```python
execution_time = timeit.timeit(lambda: solution(1041), number=10000)
print("Resultado:", solution(1041))
print("Tiempo de ejecución (10000 ejecuciones):", execution_time, "segundos")
```
